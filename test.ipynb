```python
from classifier import *

dataset = "flowers102"  # flowers102 | oxfordiiitpet
```

```python
#bitfit 

import os


model_path_flowers = "src/dino_v3/logs/dinov3_classifier_bitfit/version_0/checkpoints/best-09-2.6294.ckpt"
model_path_oxfordiiitpet = "src/dino_v3/logs/dinov3_classifier_bitfit/version_1/checkpoints/best-09-0.3587.ckpt"
method = "bitfit"  # full | lp | lora | bitfit | ncm

image_folder = "test_folder"  # test/flowers102 | test/oxfordiiitpet

for a in os.listdir(image_folder):
    dataset = a

    print(a)

    if dataset == "flowers102":
        model_path = model_path_flowers
    elif dataset == "oxfordiiitpet":
        model_path = model_path_oxfordiiitpet

    image_path = os.path.join(image_folder, a)

    for b in os.listdir(image_path):
        image_path_1 = os.path.join(image_path, b)
        print(f"Classifying image: {image_path}")
        
        results = classify_image(
            model_name=model_path,                  # or a .ckpt / prototypes.npz / run dir
            dataset=dataset,              # flowers102 | oxfordiiitpet
            image_path=f"{image_path_1}",  # path to the image
            method=method,                         # full | lp | lora | bitfit | ncm
            top_k=1,
        )

        for row in results:
            print(f"{row['rank']}. {row['class_name']:<28}")

            print("==========================================")
```

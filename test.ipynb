```python
#bitfit 

import os

image_folder = "test_folder"  # test/flowers102 | test/oxfordiiitpet

for method in ["full", "lp", "lora", "bitfit", "ncm"]:
    print(f"=============================== {method} ===============================")

    model_path_flowers = f"logs/dino_v3/flowers102/dinov3_classifier_{method}/version_0/checkpoints/best.ckpt"
    model_path_oxfordiiitpet = f"logs/dino_v3/oxfordiiitpet/dinov3_classifier_{method}/version_0/checkpoints/best.ckpt"

    for a in os.listdir(image_folder):
        dataset = a

        print(f"Processing dataset: {dataset}")

        if dataset == "flowers102":
            model_path = model_path_flowers

        elif dataset == "oxfordiiitpet":
            model_path = model_path_oxfordiiitpet

        image_path = os.path.join(image_folder, a)

        for b in os.listdir(image_path):
            image_path_1 = os.path.join(image_path, b)
            
            results = classify_image(
                model_name=model_path,
                dataset=dataset,            
                image_path=f"{image_path_1}",  
                method=method,                        
                top_k=1,
            )

            row = results[0]
            print(f"Image name : {b} -> {row['rank']}. {row['class_name']:<28}")

    print("==========================================")
```
